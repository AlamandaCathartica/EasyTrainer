# 🧊 Build Offline Bundle (run once, while online)

Freezes **every** dependency (exact wheels incl. transitive ones, sd-scripts source at a fixed commit, wrapper scripts, aria2 .deb) into your Google Drive. After this, the trainers install with **no network** and can't break from upstream changes.

**Before running:** upload these files to `MyDrive/Loras/_offline_bundle/`: `offline_bundle.py`, `train_network_wrapper.py`, `train_network_xl_wrapper.py`, `dracula.py`.

Use the same Colab runtime type (GPU, same Python) you train on. The bundle records the Python version and refuses to install on a different one.

```
from google.colab import drive
drive.mount('/content/drive')
import sys
BUNDLE_ROOT = '/content/drive/MyDrive/Loras/_offline_bundle'
sys.path.insert(0, BUNDLE_ROOT)
import offline_bundle as ob
print('Python:', ob._py_tag())
```

```
#@markdown Which trainer(s) to freeze
sd15 = True #@param {type:'boolean'}
sd15_es = False #@param {type:'boolean'}
xl_legacy = False #@param {type:'boolean'}
force_rebuild = False #@param {type:'boolean'}

for name, on in (('sd15', sd15), ('sd15_es', sd15_es), ('xl_legacy', xl_legacy)):
  if on:
    ob.build(name, root=BUNDLE_ROOT, force=force_rebuild)
```

```
# Verify checksums of everything that was frozen
for name in ('sd15', 'sd15_es', 'xl_legacy'):
  try:
    ob.verify(name, root=BUNDLE_ROOT); print(name, '✅ OK')
  except FileNotFoundError:
    print(name, '— not built')
```
